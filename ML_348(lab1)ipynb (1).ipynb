```python
sentence2 = "Data Science and Machine Learning use NumPy arrays every single day"

tokens2 = sentence2.lower().split()

print(tokens2)
```

```python
tokens2_array = np.array(tokens2)

vocabulary2 = np.unique(tokens2_array)

print("Vocabulary:", vocabulary2)
print("Vocabulary size:", len(vocabulary2))
```

The vocabulary contains 11 unique words.

```python
import numpy as np

def bag_of_words_vector(tokens, vocabulary):
    # Count how many times each vocabulary word appears
    vector = np.array([
        np.sum(np.array(tokens) == word)
        for word in vocabulary
    ])

    return vector
```

```python
vec2 = bag_of_words_vector(tokens2, vocabulary2)

print("Vocabulary:", vocabulary2)
print("Bag-of-Words vector:", vec2)
```

```python
sentence_1 = "machine learning uses data"
sentence_2 = "machine learning uses algorithms"

tokens_1 = sentence_1.lower().split()
tokens_2 = sentence_2.lower().split()


vocab_1 = np.unique(np.array(tokens_1 + tokens_2))


vec_1 = bag_of_words_vector(tokens_1, vocab_1)
vec_2 = bag_of_words_vector(tokens_2, vocab_1)

similarity_1 = np.dot(vec_1, vec_2) / (
    np.linalg.norm(vec_1) * np.linalg.norm(vec_2)
)

print("Pair 1 similarity:", similarity_1)
```

Cosine similarity tells us how much two vectors point in the same direction. Similar sentences usually have many common words, so their vectors point in similar directions and the value is close to 1. Unrelated sentences have fewer or no common words, so their vectors point in different directions and the value is close to 0. A cosine similarity of exactly 1 means the two vectors point in exactly the same direction, meaning they are perfectly similar in their vector representation.  AP24110011348


```python
paragraph = """
Machine learning is changing technology.
Machine learning helps computers learn from data.
Data is important for building machine learning models.
Learning from data makes models more useful.
"""


tokens = paragraph.lower().split()


tokens_array = np.array(tokens)
vocabulary = np.unique(tokens_array)


vec = bag_of_words_vector(tokens, vocabulary)


sorted_indices = np.argsort(vec)[::-1]

top_3_indices = sorted_indices[:3]

print("Top 3 most frequent words:")

for i in top_3_indices:
    print(vocabulary[i], ":", vec[i])
```

```python
import numpy as np



sentence_1 = "machine learning uses data to build models"
sentence_2 = "machine learning uses data to train models"
sentence_3 = "football players practice every day"



tokens_1 = sentence_1.lower().split()
tokens_2 = sentence_2.lower().split()
tokens_3 = sentence_3.lower().split()




combined_tokens = tokens_1 + tokens_2 + tokens_3

shared_vocabulary = np.unique(np.array(combined_tokens))

print("Shared Vocabulary:")
print(shared_vocabulary)



vec_1 = bag_of_words_vector(tokens_1, shared_vocabulary)
vec_2 = bag_of_words_vector(tokens_2, shared_vocabulary)
vec_3 = bag_of_words_vector(tokens_3, shared_vocabulary)

print("\nVector 1:", vec_1)
print("Vector 2:", vec_2)
print("Vector 3:", vec_3)




def cosine_similarity(A, B):
    return np.dot(A, B) / (
        np.linalg.norm(A) * np.linalg.norm(B)
    )



similarity_12 = cosine_similarity(vec_1, vec_2)
similarity_13 = cosine_similarity(vec_1, vec_3)
similarity_23 = cosine_similarity(vec_2, vec_3)

print("\n===== COSINE SIMILARITY =====")
print("Sentence 1 & Sentence 2:", similarity_12)
print("Sentence 1 & Sentence 3:", similarity_13)
print("Sentence 2 & Sentence 3:", similarity_23)




similarities = {
    "Sentence 1 & Sentence 2": similarity_12,
    "Sentence 1 & Sentence 3": similarity_13,
    "Sentence 2 & Sentence 3": similarity_23
}

most_similar_pair = max(similarities, key=similarities.get)

print("\nMost similar pair:", most_similar_pair)
print("Similarity score:", similarities[most_similar_pair])
```

1. Why do we need to convert text into numbers?

Machine learning models work with numerical data rather than raw words or sentences. Converting text into numbers allows the model to perform mathematical operations and find patterns in the text.

2. What information does Bag-of-Words lose?

Bag-of-Words loses the order of words and the structure of the sentence. It only records how many times each word appears. For example, "dog bites man" and "man bites dog" could have the same Bag-of-Words vector even though their meanings are different.

3. Why is cosine similarity better than raw word-count differences?

Cosine similarity compares the direction of the vectors rather than just their size. This makes it useful when sentences have different lengths. Two sentences can have similar word patterns even if one contains more words than the other.

4. Real-world application

Plagiarism detection is one application. A system can convert documents into numerical vectors and compare their similarity. If two documents have very similar vector representations, the system can flag them for further checking.
